## The experiment in the book

In Lab 26 you drew a clock face from memory and then looked at clock drawings by patients with
right-parietal damage, who cram all twelve numbers into the right half of the circle and leave
the left half blank. This is **hemispatial neglect**: not blindness, but an attentional deficit
-- the left side of space is no longer a place the patient's attention system can be directed
to, even though their eyes and visual cortex are intact.

## Record your results

This Lab is mostly observational: healthy attention cannot be switched off the way neglect

```python
# example data: replace with your own counts from Steps 1 and 2
clock_check = pd.Series({"numbers in left half of your clock": 6,
                          "numbers in right half of your clock": 6}, name="count")
display(clock_check.to_frame())

# example data: replace with your own word recall from Step 4's reading exercise
reading_check = pd.Series({"words recalled from the attended (right) side": 5,
                            "words recalled from the unattended (left) side": 2}, name="words recalled")
display(reading_check.to_frame())
```

## Compare

The book does not give quantitative data on how a healthy reader performs on either check, so
there is no published number to replicate here -- only the qualitative *pattern* described in
the text: a patient's clock has all **12** numbers on the right and **none** on the left,
while a healthy clock should be roughly even.

```python
left, right = clock_check["numbers in left half of your clock"], clock_check["numbers in right half of your clock"]
print(f"Your clock: {left} numbers on the left, {right} on the right (12 total).")
print("Patients with hemispatial neglect: 0 on the left, 12 on the right (all crammed into one half).")
if abs(left - right) <= 2:
    print("Your clock is roughly even, as expected for intact spatial attention -- the contrast with")
    print("the all-or-nothing patient pattern is the point, not a numeric match.")
else:
    print("Your clock is noticeably lopsided. Before concluding anything, check whether you simply")
    print("ran out of room while drawing, rather than a genuine attentional effect.")
```
