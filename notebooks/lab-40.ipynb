## The experiment in the book

In Lab 40 you and a friend assigned ten simple actions to one of three conditions by die roll
— perform it, watch your friend perform it, or imagine it — recording the hidden assignment
before testing, then, 24 hours later, tried to recall which condition each action belonged
to. Some imagined actions get confused with performed actions, especially after vivid or
repeated imagining, because imagining an action recruits some of the same sensory machinery
as actually doing it.

## Record your results

This Lab does not reduce to a single accuracy percentage reported in the book, so keep a
structured log instead: one row per action, its true condition, and what your friend
recalled 24 hours later. The imagined-versus-performed confusions are the interesting cells.

```python
# example data: replace with your own 24-hour follow-up test
actions = ["Touch left ear", "Stand up and sit down", "Clap three times", "Wave right hand",
           "Tap the table", "Close eyes for five seconds", "Cross arms", "Point to the ceiling",
           "Snap fingers", "Nod head"]
true_condition = ["performed", "imagined", "watched", "performed", "imagined",
                   "watched", "performed", "imagined", "watched", "performed"]      # example data: from your die rolls
recalled_condition = ["performed", "performed", "watched", "performed", "imagined",
                       "imagined", "performed", "performed", "watched", "performed"]  # example data: your friend's 24h answers

log = pd.DataFrame({"action": actions, "true": true_condition, "recalled": recalled_condition})
log["correct"] = log["true"] == log["recalled"]
log["error type"] = np.where(log["correct"], "-", log["true"] + " -> recalled as " + log["recalled"])
display(log)

print(f"Overall accuracy: {log['correct'].mean():.0%}")
imagined_rows = log[log["true"] == "imagined"]
if len(imagined_rows):
    imagined_as_performed = (imagined_rows["recalled"] == "performed").mean()
    print(f"Of the imagined actions, the proportion later recalled as performed: {imagined_as_performed:.0%}")

errors = log.loc[~log["correct"], "error type"]
if len(errors):
    errors.value_counts().plot.barh(figsize=(6, 2.5), color="firebrick")
    plt.xlabel("count"); plt.tight_layout(); plt.show()
else:
    print("No errors this run -- try a longer action list, or a longer delay, to see the effect emerge.")
```
