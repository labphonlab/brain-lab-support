## The experiment in the book

In Lab 28 you wrote out a number-to-letter and number-plus-3 sequence three times: once
alternating between the two rules item by item, and once doing each rule in a single unbroken
run. The alternating run took substantially longer than the two separate runs added together,
even though neither task alone was hard. That extra time is the **task-switching cost** --
the price of reconfiguring the brain's rules for what to do next, not of the tasks themselves.

## Record your results

Time all three runs from Steps 1-2 with a stopwatch: the alternating run, the all-letters
run, and the all-numbers-plus-3 run.

```python
n_items = 26
my_times_s = {"alternating (Step 1)": 74, "letters only (Step 2)": 22,
              "numbers+3 only (Step 2)": 24}  # example data: replace with yours

results = pd.Series(my_times_s, name="time (s)")
display(results.to_frame())
(results * 1000 / n_items).plot.barh(color=["firebrick", "steelblue", "steelblue"], figsize=(6, 2.6))
plt.xlabel("milliseconds per item"); plt.gca().invert_yaxis(); plt.show()
```

## Compare

The book reports that the alternating time is "often 30-50% longer" than the sum of the two
separate-task times. Add your two separate runs together and compare with your alternating
run.

```python
separate_total = results["letters only (Step 2)"] + results["numbers+3 only (Step 2)"]
alternating = results["alternating (Step 1)"]
switch_cost_pct = (alternating - separate_total) / separate_total * 100

print(f"Sum of separate-task times: {separate_total:.0f} s")
print(f"Alternating-task time:      {alternating:.0f} s")
print(f"Your switch cost: {switch_cost_pct:.0f}% longer than the two separate runs combined "
      "(book's typical range: 30-50%)")
if 30 <= switch_cost_pct <= 50:
    print("Inside the typical 30-50% range reported in Lab 28.")
elif switch_cost_pct > 0:
    print("A positive switch cost, as predicted, though outside the 30-50% range quoted in the book --")
    print("handwriting speed, practice with the two individual tasks, and list length all affect the size.")
else:
    print("No positive switch cost in your data. Check whether you paused between items in the")
    print("alternating run, or whether practice on the separate runs (done second) made them faster.")
```

## Pool the class data

One run of each condition is a single noisy measurement. Pool the class: each person adds one
row (an anonymous ID and their switch-cost percentage), downloads the sheet as CSV, and loads
it here. The rows below are placeholders so the cell runs.

```python
import io
csv_text = """id,switch_cost_pct
P01,42
P02,55
P03,28
P04,38
P05,47
"""  # example data: replace with your class CSV
df = pd.read_csv(io.StringIO(csv_text))
boot = [df.switch_cost_pct.sample(len(df), replace=True).mean() for _ in range(5000)]
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f"Class mean switch cost: {df.switch_cost_pct.mean():.0f}% "
      f"(95% bootstrap interval {lo:.0f}-{hi:.0f}%, n = {len(df)})")
df.switch_cost_pct.plot.hist(bins=8, color="firebrick", alpha=0.7, figsize=(6, 3))
plt.axvspan(30, 50, color="grey", alpha=0.2, label="typical range (Lab 28)")
plt.xlabel("switch cost (%)"); plt.legend(); plt.show()
```

## Questions to think about
