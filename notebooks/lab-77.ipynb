## The experiment in the book

In Lab 77 you replayed several memories and noticed their **visual perspective**: first-person (seeing
the scene through your own eyes) or third-person (watching yourself from outside, as if in a film). You
found you could deliberately switch perspective, and that recent memories tend towards first-person while
older memories tend towards third-person (Nigro & Neisser, 1983). Emotion cuts across this in a specific
way: self-conscious moments (embarrassment, being watched) tend towards third-person, while memories
focused on the raw feeling itself tend to stay first-person and carry more of the original emotional charge
(Berntsen & Rubin, 2006).

## Record your results

Log each memory you replayed: roughly how old it is, whether it involved physical action, whether it
carried strong emotion, and which perspective it came in. Use "1" for first-person and "3" for
third-person. Four or five memories is enough to see your own pattern.

```python
import pandas as pd

# example data: replace with your own memories
memories = pd.DataFrame([
    {"memory": "cooking dinner last night",      "age_group": "recent",    "physical": True,  "emotional": False, "perspective": 1},
    {"memory": "walking to class this morning",  "age_group": "recent",    "physical": True,  "emotional": False, "perspective": 1},
    {"memory": "an embarrassing moment at school","age_group": "childhood","physical": False, "emotional": True,  "perspective": 3},
    {"memory": "riding a bicycle as a child",     "age_group": "childhood", "physical": True,  "emotional": False, "perspective": 1},
    {"memory": "a birthday party, age 7",         "age_group": "childhood", "physical": False, "emotional": True,  "perspective": 3},
])  # example data: replace with your own
memories["perspective_label"] = memories["perspective"].map({1: "first-person", 3: "third-person"})
display(memories)

counts = memories.groupby(["age_group", "perspective_label"]).size().unstack(fill_value=0)
counts.plot.bar(figsize=(6, 3.2), color=["steelblue", "firebrick"])
plt.ylabel("number of memories"); plt.title("Perspective by memory age"); plt.xticks(rotation=0)
plt.show()
```

## Compare

Lab 77 reports that "recent memories tend to be first-person (field), while older memories tend to
shift to third-person (observer)" (Nigro & Neisser, 1983). The book does not give an exact proportion,
so compare the *direction* of your own data rather than a specific number: is the share of third-person
memories higher among your childhood memories than among your recent ones?

```python
share_third = memories.groupby("age_group")["perspective"].apply(lambda s: (s == 3).mean())
display(share_third.rename("share third-person"))

if share_third.get("childhood", 0) > share_third.get("recent", 0):
    print("Your pattern matches the book: more third-person perspective among older memories.")
elif share_third.get("childhood", 0) == share_third.get("recent", 0):
    print("No difference in this small sample -- with only a handful of memories, that is unsurprising.")
else:
    print("Your pattern runs the other way. With only a handful of memories logged, and a self-selected")
    print("sample of what you happened to think of, this is not strong evidence against the finding --")
    print("it just means your own small set of memories doesn't show it.")
```
