## The experiment in the book

In Lab 56 you interlaced a fork warmed in about 40 degrees C water with a fork cooled in about
20 degrees C water, then pressed the alternating warm and cool tines against your forearm. Even
though neither temperature alone was uncomfortable, many people feel a distinct burning or
stinging sensation from the combination -- the thermal grill illusion, first reported by
Thunberg in 1896 and explained by Craig and Bushnell (1994) as pain manufactured centrally from
two harmless thermal signals, with the exact circuitry still under debate (Fardo et al., 2020).

## Record your results

Rate each condition from 0 (no sensation) to 10 (intense burning/pain), immediately after
pressing the fork(s) against your forearm for a few seconds: warm fork alone, cool fork alone,
and the interlaced warm-cool grill. Repeat the sequence twice, waiting a minute or two between
repeats so your skin returns to its resting temperature. Stay within the safety limits above
throughout.

```python
log = pd.DataFrame({
    "repeat": [1, 1, 1, 2, 2, 2],
    "condition": ["warm alone", "cool alone", "interlaced grill"] * 2,
    "rating (0-10)": [1, 1, 6, 0, 1, 5],   # example data: replace with your own ratings
})
display(log)
summary = log.groupby("condition")["rating (0-10)"].mean().reindex(["warm alone", "cool alone", "interlaced grill"])
display(summary.to_frame("mean rating (0-10)"))
summary.plot.barh(color=["firebrick", "steelblue", "purple"], figsize=(6, 2.4))
plt.xlim(0, 10); plt.xlabel("sensation rating (0-10)"); plt.show()
print("The book's claim here is qualitative, not a numeric scale: the interlaced grill should feel far more")
print("unpleasant than either temperature alone, even though neither alone is painful. Check whether your")
print("own ratings show that shape.")
```

## Challenge

Vary the temperature gap between the two forks (for example, 35 degrees C and 25 degrees C, well
within the safety limits above) and rate the burning sensation again. See whether a smaller
temperature gap produces a weaker illusion than the original roughly 20-degree gap.

```python
# Example data: replace with your own ratings, keeping to the safety limits above (never above
# about 42 degrees C).
temperature_gap_c = [5, 10, 20]   # example: (warm C) - (cool C), always within safe limits
burning_rating = [1, 3, 6]        # example data: 0 (no burning) to 10 (intense)

results = pd.DataFrame({"temperature gap (C)": temperature_gap_c, "burning rating (0-10)": burning_rating})
display(results)
results.plot.line(x="temperature gap (C)", y="burning rating (0-10)", marker="o", legend=False, figsize=(6, 3))
plt.ylim(0, 10); plt.ylabel("burning rating (0-10)"); plt.show()
print("If the rating rises with the temperature gap, this suggests the illusion scales with how far apart the")
print("two thermal signals are, not just their alternating pattern -- but always keep the warm side within")
print("the safety limits from the Lab, however large a gap you are curious about.")
```
