## The experiment in the book

In Lab 11 you looked away from a ticking clock, then glanced quickly back at the second
hand. The first tick after your glance seemed to last far longer than the ticks that
followed, as if the clock had briefly frozen. This is **chronostasis**: during the saccade
(the fast eye movement itself) the brain suppresses vision, then backdates the first stable
image it sees to paper over the gap, stretching its apparent duration.

## Record your results

This illusion is measured in milliseconds in the lab, using eye-tracking equipment that
can time saccades precisely. A stopwatch or phone timer cannot resolve that scale, so
instead of timing the stretch, rate it: each time you glance at the clock, judge how
stretched the first tick felt, from 1 (not stretched at all) to 5 (felt frozen). Run a few
trials glancing from across the room ("far") and from close to the clock ("near"), plus a
few control trials where you were already looking at the clock before it ticked
("no saccade"). Type your ratings below.

```python
# example data: replace with your own ratings (1 = not stretched, 5 = felt frozen)
trials = pd.DataFrame({
    "condition": ["far", "far", "far", "near", "near", "near", "no saccade", "no saccade", "no saccade"],
    "rating":    [4,     5,     4,     3,      2,      3,      1,            1,            2],
})
display(trials)
trials.groupby("condition")["rating"].mean().reindex(["far", "near", "no saccade"]).plot.bar(
    color=["firebrick", "steelblue", "grey"], figsize=(5, 3))
plt.ylabel("mean stretch rating (1-5)"); plt.ylim(0, 5); plt.xticks(rotation=0); plt.show()
```
