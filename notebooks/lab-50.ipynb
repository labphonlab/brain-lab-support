## The experiment in the book

In Lab 50 you used a mirror to make your right hand's reflection appear where your hidden left
hand would be, then moved your right hand while watching the reflection as if it were your left.
The book explains that this technique, developed to treat phantom limb pain, works by letting
vision override a stuck or absent proprioceptive signal and update the brain's body model --
though a review of properly placebo-controlled trials found the clinical benefit less certain
than early reports suggested.

## Challenge

Try the mirror set-up with a task that requires coordination, such as mirror-drawing (tracing a
simple shape while looking only at its reflection). Time how long the mirrored version takes
compared with drawing the same shape directly, and see if the gap shrinks with practice.

```python
# Example data: replace with your own timings for tracing a shape directly vs. via the mirror.
attempt = [1, 2, 3, 4]
mirror_time_s = [42, 35, 28, 24]    # example data: seconds to trace the shape looking only at the mirror
direct_time_s = [9, 8, 9, 8]        # example data: seconds to trace the same shape looking directly at your hand

results = pd.DataFrame({"attempt": attempt, "mirror (s)": mirror_time_s, "direct (s)": direct_time_s})
display(results)
results.plot.line(x="attempt", y=["mirror (s)", "direct (s)"], marker="o", figsize=(6, 3))
plt.ylabel("time to trace shape (s)"); plt.show()
print("If the mirror times fall towards the direct times with practice, that is the visuomotor system")
print("recalibrating to the reversed mapping -- a small-scale version of the body-model updating that")
print("mirror therapy relies on.")
```
