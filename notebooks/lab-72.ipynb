## The AI side

The book's AI Connection describes **swarm intelligence**: simple units, coupled by local
rules, that synchronise without any central conductor. The **Kuramoto model** (already
referenced in this Lab through Strogatz, 2003) is the classic mathematical version of exactly
that idea. Below, twenty independent oscillators start at random phases and lock together
using nothing but a shared coupling term.

*This part downloads a small open model or library, so it runs in Colab (it may take a minute the first time).*

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
n_oscillators = 20
coupling_K = 2.0
natural_freqs = np.random.normal(1.0, 0.3, n_oscillators)
phases = np.random.uniform(0, 2 * np.pi, n_oscillators)
dt, steps = 0.05, 300
order_over_time = []

for _ in range(steps):
    mean_field = np.mean(np.exp(1j * phases))
    r, psi = np.abs(mean_field), np.angle(mean_field)
    phases = phases + dt * (natural_freqs + coupling_K * r * np.sin(psi - phases))
    order_over_time.append(r)

plt.figure(figsize=(6, 3.5))
plt.plot(order_over_time)
plt.xlabel("time step"); plt.ylabel("synchrony (order parameter r: 0 = random, 1 = in phase)")
plt.title("Kuramoto model: spontaneous phase-locking from random starts")
plt.show()
print(f"Synchrony after {steps} steps: r = {order_over_time[-1]:.2f}",
      f"(started at r = {order_over_time[0]:.2f})")
```

This is a model of abstract phase oscillators, coupled through one clean shared "mean field"
term and one number for coupling strength. Real clapping hands are coupled through sound
that arrives with a delay and through vision, not through a single mathematical coupling
term, and real neurons are not simple oscillators reducible to one phase value each. The
Kuramoto model captures the qualitative shape of the phenomenon -- many simple units syncing
without a leader -- not the acoustic or biological mechanism your own ears and hands actually
use.
