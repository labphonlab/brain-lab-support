## The experiment in the book

In Lab 49 you stepped onto an escalator or moving walkway that had been switched off, walking at
your normal pace. Even though you could see it was stationary, most people feel a brief,
involuntary lurch as their foot lands, as if the missing movement still needed to be corrected
for. The book explains this as a context-specific motor prediction ("escalators move") being
violated by a stationary surface, and notes that the lurch shrinks with repeated trials on the
same escalator.

## Record your results

A lurch is hard to judge from the inside, so use the slow-motion video method the book
suggests: film yourself (Step 4's smartphone version) stepping onto the stopped escalator or
off the treadmill, then watch each clip in slow motion. Rate how large the lurch or wobble
looks, from 0 (no visible reaction) to 5 (an obvious stumble), and repeat on the same escalator
or treadmill a few times if you can.

```python
trials = pd.DataFrame({
    "trial number": [1, 2, 3, 4],
    "lurch rating (0-5)": [4, 3, 2, 1],   # example data: replace with your own slow-motion video ratings
})
display(trials)
trials.plot.line(x="trial number", y="lurch rating (0-5)", marker="o", legend=False, figsize=(6, 3))
plt.ylim(0, 5); plt.ylabel("lurch rating (0-5)"); plt.show()
```

## Compare

The book reports that the lurch does not stay constant: across repeated trials on the same
stopped escalator, it shrinks noticeably as the motor system revises its prediction (Bunday et
al., 2006). Check whether your own ratings show the same downward trend.

```python
import numpy as np

slope = np.polyfit(trials["trial number"], trials["lurch rating (0-5)"], 1)[0]
print(f"Slope across your trials: {slope:+.2f} rating points per trial")
if slope < 0:
    print("Your lurch shrank across trials, matching the book's report that the motor prior is revised with")
    print("repeated exposure to the same stopped escalator.")
else:
    print("Your lurch did not shrink here. A handful of trials is a small sample, and the effect Bunday et")
    print("al. report is measured over many more repetitions with instrumented sensors, not eyeball ratings")
    print("from a phone video, so treat this as a rough check rather than a strict test.")
```
