## The experiment in the book

In Lab 02 you watched a rotating hollow (concave) mask -- as a physical mask, a video, or a
folded-paper version -- and found that the concave side stubbornly looked like an ordinary,
convex face that seemed to follow you as you moved. The illusion persisted even though you knew
the mask was hollow, because your brain's lifelong prior that faces bulge outward overrode the
shading and perspective cues that correctly signalled a concave surface.

## Record your results

Watch the rotating mask (Option B), a physical mask (Option A), or the paper fold (Option C)
for at least five separate viewings -- five loops of the video, or five approach-and-recede
cycles. Each time, note whether you perceived a normal convex face or correctly saw the
concave (hollow) surface.

```python
# example data: replace 1 with 0 for any viewing where you correctly saw the concave/hollow surface
my_trials = pd.DataFrame({
    "viewing": [1, 2, 3, 4, 5],
    "saw_convex_face": [1, 1, 1, 1, 1],   # example data: 1 = illusion (looked convex), 0 = correctly seen as concave
})
display(my_trials)
convex_rate = my_trials["saw_convex_face"].mean()
print(f"You perceived a normal convex face on {convex_rate:.0%} of viewings.")
```

## Compare

Lab 02 reports a striking contrast from one brain-imaging study: healthy volunteers
misclassified the hollow mask as a normal convex face about 99% of the time, compared with
only about 6% for patients with schizophrenia in the same study (Dima et al., 2009) --
evidence that the illusion depends on unusually strong top-down prediction.

```python
print(f"Your convex-illusion rate: {convex_rate:.0%}")
print("Healthy volunteers in Dima et al. (2009): about 99%")
print("Patients with schizophrenia in the same study: about 6%")
if convex_rate >= 0.8:
    print("Close to the healthy-volunteer figure: your top-down 'faces are convex' prior is")
    print("doing what it does for almost everyone tested.")
else:
    print("Lower than the healthy-volunteer figure. This is a single person doing a handful of")
    print("informal viewings, not a controlled lab task, so many things other than the group")
    print("difference reported in the study could explain a lower rate -- for example, already")
    print("knowing the mask is hollow, viewing it very close up, or using only one eye.")
```
