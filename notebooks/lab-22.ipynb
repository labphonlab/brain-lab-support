## The experiment in the book

In Lab 22 you showed a friend the Simons and Chabris (1999) selective-attention video and
asked them to count passes made by the white team. About half of first-time viewers fail to
notice a person in a gorilla suit walking through the middle of the scene for about nine
seconds, beating its chest. This is **inattentional blindness**: the gorilla is on the retina,
but attention is fully committed to the counting task, so it never reaches awareness.

## Record your results

Each person can only be tested once -- once they know about the gorilla, they will look for
it. Build up your own sample over time: every time you find a new, genuinely naive viewer,
show them the video, ask for their pass count, then ask "did you notice anything unusual?"
and log whether they saw the gorilla, without telling future testees what to expect.

```python
# example data: replace with your own log as you test more naive viewers
trials = pd.DataFrame([
    {"id": "T01", "reported_passes": 15, "correct_passes": 15, "noticed_gorilla": False},
    {"id": "T02", "reported_passes": 14, "correct_passes": 15, "noticed_gorilla": True},
    {"id": "T03", "reported_passes": 15, "correct_passes": 15, "noticed_gorilla": False},
    {"id": "T04", "reported_passes": 16, "correct_passes": 15, "noticed_gorilla": False},
])
display(trials)
trials.noticed_gorilla.value_counts().plot.bar(color=["firebrick", "steelblue"], figsize=(4, 3))
plt.ylabel("number of viewers"); plt.xticks(rotation=0); plt.show()
```

## Compare

Simons and Chabris (1999) found that about **50%** of viewers who count passes fail to
notice the gorilla. With so few testees, your own miss rate will be a rough estimate --
every extra person you test narrows it.

```python
miss_rate = 1 - trials.noticed_gorilla.mean()
n = len(trials)
print(f"Your miss rate: {miss_rate:.0%} of {n} viewers (Simons & Chabris, 1999: about 50%)")
if n < 8:
    print("With so few viewers, a single extra person can swing this a lot; the published 50% figure"
          " comes from a much larger sample.")
elif abs(miss_rate - 0.5) <= 0.2:
    print("Reasonably close to the classic ~50% miss rate.")
else:
    print("Some way from the classic ~50% figure -- worth checking whether your testees were really"
          " naive, and whether they were counting carefully rather than just watching.")
```
