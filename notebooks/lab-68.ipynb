## The experiment in the book

In Lab 68 you recorded a five-minute conversation with a friend and watched it back for
**postural mirroring** — moments when one person crosses their legs, leans forward, or
touches their face, and the other unconsciously copies it within a few seconds. In Step 3,
your friend secretly chose one posture to adopt at intervals, without telling you which one,
and a third person who did not know which posture was chosen watched the recording and coded
whether you matched it. This blind check matters: if you already knew which posture to look
for, you could not tell genuine unconscious mimicry from something you consciously noticed and
copied. Chartrand and Bargh (1999) called this the **chameleon effect**.

## Record your results

Tally mirrored instances minute by minute from the recording, then note whether your blind
coder marked the Step 3 posture as copied, and how long it took to appear.

```python
# example data: replace with yours
mimicry_log = pd.DataFrame({
    "minute": [1, 2, 3, 4, 5],
    "mirrored_instances_observed": [0, 1, 2, 1, 3],
})
display(mimicry_log)
mimicry_log.plot(x="minute", y="mirrored_instances_observed", marker="o",
                  legend=False, figsize=(5, 3.5))
plt.ylabel("mirrored instances per minute"); plt.xlabel("minute of conversation"); plt.show()
print(f"Total mirrored instances in the conversation: {mimicry_log.mirrored_instances_observed.sum()}")

blind_coder_marked_posture_mirrored = True   # example data: replace with yours
seconds_to_mirror = 12                        # example data: replace with yours
print(f"Step 3 posture marked as mirrored by blind coder: "
      f"{blind_coder_marked_posture_mirrored} (after {seconds_to_mirror} s)")
```

## Compare

Lab 68 does not give a numeric mimicry rate to match — and it says so explicitly: direct

informal check rather than a strict test.

```python
half = len(mimicry_log) // 2
first_half = mimicry_log.iloc[:half].mirrored_instances_observed.mean()
second_half = mimicry_log.iloc[half:].mirrored_instances_observed.mean()
print(f"Mirrored instances, first half of the conversation: {first_half:.1f} per minute")
print(f"Mirrored instances, second half of the conversation: {second_half:.1f} per minute")
print("An increase across the conversation would be consistent with mimicry building as")
print("rapport develops, but Lab 68 is explicit that the strength of this effect is debated,")
print("so a flat or noisy pattern from five minutes of one conversation is not surprising.")
```

## The AI side

The book compares mimicry with **preference fine-tuning**: adjusting a language model's
outputs to match what humans want to see (Ouyang et al., 2022, introduced the reinforcement
learning from human feedback recipe now used across the industry). The demonstration below
uses a related but simpler technique, supervised instruction-tuning, because both small
models below can run without any extra downloads.

*This part downloads a small open model or library, so it runs in Colab (it may take a minute the first time).*

```python
from transformers import pipeline

base = pipeline("text2text-generation", model="t5-small")
tuned = pipeline("text2text-generation", model="google/flan-t5-small")

prompt = "How do I make a new friend feel welcome at a party?"
print("t5-small (pretrained on raw text, never tuned to follow instructions):")
print(" ", base(prompt, max_new_tokens=40)[0]["generated_text"])
print()
print("flan-t5-small (fine-tuned on human-written instruction-and-answer examples):")
print(" ", tuned(prompt, max_new_tokens=40)[0]["generated_text"])
```

flan-t5-small was tuned by supervised fine-tuning on human-written instruction-and-answer
pairs, not the full reinforcement-learning pipeline described in the book — but both share
the same goal of pulling a model's raw output toward what humans prefer. The deeper
disanalogy remains: t5-small did not choose to ignore your question, and flan-t5-small did
not choose to be helpful. Both are weights adjusted by gradient descent on a fixed training
objective, with none of the social motivation that makes a person unconsciously mirror a
conversation partner they like.
