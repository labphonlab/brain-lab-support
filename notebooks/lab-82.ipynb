## The experiment in the book

In Lab 82 you tried to describe the subjective "raw feel" of seeing red -- not its wavelength, not its
associations, just the quale itself -- and found it essentially impossible to put into words. You compared
descriptions with a friend and realised that shared vocabulary ("warm", "bright") does not prove shared
experience: you have no way to check whether your red is their red.

## Record your results

This is not a measurable experiment -- that is the point of the Lab. Still, a structured log of what you
and your friend actually said can reveal something: how much overlap is there in the words you each
reach for, and how quickly do both of you give up on describing the "raw feel" and fall back on causes or
associations instead?

```python
import pandas as pd

# example data: replace with the actual words you and your friend used
words = pd.DataFrame([
    {"person": "you",    "word": "warm",      "type": "quality"},
    {"person": "you",    "word": "intense",   "type": "quality"},
    {"person": "you",    "word": "fire",      "type": "association"},
    {"person": "friend", "word": "warm",      "type": "quality"},
    {"person": "friend", "word": "loud",      "type": "quality"},
    {"person": "friend", "word": "stop sign", "type": "association"},
])  # example data: replace with your own session's words
display(words)

shared = set(words[words.person == "you"]["word"]) & set(words[words.person == "friend"]["word"])
print(f"Words you both used: {sorted(shared) or 'none'}")
words.groupby(["person", "type"]).size().unstack(fill_value=0).plot.bar(figsize=(6, 3))
plt.ylabel("number of words used")
plt.title("Quality words vs. association words, by person")
plt.xticks(rotation=0)
plt.show()
```
