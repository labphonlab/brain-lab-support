## The experiment in the book

In Lab 23 you watched a rapid stream of letters (about ten per second) and tried to identify
two coloured targets. When the two targets were 200-500 milliseconds apart, you reliably
caught the first but often missed the second -- the **attentional blink**. Targets very close
together, or more than about 500 ms apart, were both caught reliably. Trying harder did not
help: the blink is a structural bottleneck, not a lapse of motivation.

## Record your results

Find an online RSVP attentional-blink demonstration that lets you set the lag between the
two targets. Run at least 10 trials at each of a few lags -- for example 100 ms (targets
almost adjacent), 300 ms and 450 ms (inside the reported dip), and 800 ms (well outside it)
-- and record the proportion of trials where you correctly identified the second target,
given that you caught the first.

```python
lag_ms = [100, 300, 450, 800]
t2_given_t1_accuracy = {100: 0.90, 300: 0.45, 450: 0.50, 800: 0.85}  # example data: replace with yours

results = pd.Series(t2_given_t1_accuracy, name="T2|T1 accuracy")
results.index.name = "lag between targets (ms)"
display(results.to_frame())
results.plot(marker="o", figsize=(6, 3.5), color="firebrick")
plt.xlabel("lag between T1 and T2 (ms)"); plt.ylabel("T2 accuracy given T1 correct")
plt.ylim(0, 1); plt.show()
```

## Compare

Figure 3.2's data (Raymond, Shapiro, & Arnell, 1992) show accuracy dropping in the
180-450 ms range after the first target, while accuracy stays high for targets right next
to each other or well after that window (values in the figure are accurate to about
±2 percentage points). The book's text describes the dip more broadly as the 200-500 ms
window.

```python
inside_dip = [lag for lag in lag_ms if 200 <= lag <= 500]
outside_dip = [lag for lag in lag_ms if lag not in inside_dip]

mean_inside = np.mean([t2_given_t1_accuracy[lag] for lag in inside_dip]) if inside_dip else float("nan")
mean_outside = np.mean([t2_given_t1_accuracy[lag] for lag in outside_dip]) if outside_dip else float("nan")

print(f"Your mean accuracy inside the 200-500 ms window:  {mean_inside:.0%}  (lags: {inside_dip})")
print(f"Your mean accuracy outside that window:           {mean_outside:.0%}  (lags: {outside_dip})")
if mean_inside < mean_outside:
    print("Lower accuracy inside the window, as the classic 200-500 ms attentional-blink dip predicts.")
else:
    print("No clear dip in your data. A handful of trials per lag is noisy -- try more trials, or a")
    print("demo that lets you set lags closer to 300-350 ms, where the dip is usually deepest.")
```

## Pool the class data

One person's dip is noisy at only a few trials per lag. Pool the class: each person adds one
row per lag they tested (an anonymous ID, the lag in ms, and their accuracy), downloads the

```python
import io
csv_text = """id,lag_ms,accuracy
P01,100,0.9
P01,300,0.4
P01,450,0.5
P01,800,0.8
P02,100,0.8
P02,300,0.5
P02,450,0.55
P02,800,0.9
P03,100,0.95
P03,300,0.3
P03,450,0.4
P03,800,0.85
"""  # example data: replace with your class CSV
df = pd.read_csv(io.StringIO(csv_text))
class_curve = df.groupby("lag_ms").accuracy.agg(["mean", "sem"])
display(class_curve)
plt.errorbar(class_curve.index, class_curve["mean"], yerr=class_curve["sem"], marker="o", color="firebrick")
plt.axvspan(200, 500, color="grey", alpha=0.2, label="attentional-blink window")
plt.xlabel("lag between T1 and T2 (ms)"); plt.ylabel("mean T2|T1 accuracy"); plt.ylim(0, 1)
plt.legend(); plt.show()
```
