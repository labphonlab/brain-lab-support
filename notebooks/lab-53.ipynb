## The experiment in the book

In Lab 53 you used a stick to reach, tap and explore objects for several minutes, then noticed
that your sense of touch seemed to move to the tip of the stick, and that you began to misjudge
how far you could reach without it. The book explains this with tool incorporation: after just a
few minutes of active use, neurons that code "near the hand" expand to include the space around
the tool's tip, extending the brain's body model -- an effect that appears quickly and fades
quickly once the tool is put down (Iriki et al., 1996; Maravita & Iriki, 2004).

## Record your results

There is no natural numeric measure for tool incorporation in this Lab, so log a few short
ratings instead. Before using the stick, and again after about five minutes of active use, rate
from 0 (not at all) to 5 (very strongly): how much the stick's tip feels like part of your
hand, and how confidently you can judge whether an object is within reach using only the
stick, eyes closed.

```python
log = pd.DataFrame({
    "timing": ["before use", "after ~5 min of active use"],
    "tip feels like part of hand (0-5)": [0, 3],     # example data: replace with your own ratings
    "confident reach judgement (0-5)": [2, 4],       # example data
})
display(log)
log.set_index("timing").plot.barh(figsize=(7, 2.5))
plt.xlim(0, 5); plt.xlabel("rating (0-5)"); plt.tight_layout(); plt.show()
```

## Questions to think about

If you put the stick down for ten minutes and then picked it up again, would you expect the "tip feels like part of my hand" rating to return instantly, or to build up again the way it did the first time?

## Challenge

Test the "phantom phone" effect described in the smartphone version: use your phone actively
(texting, browsing, taking photos) for about ten minutes, then put it down and immediately try
to reach for a nearby object with the same hand. Rate how strange or "empty" your hand feels, and
repeat after a period without touching your phone at all, for comparison.

```python
# Example data: replace with your own ratings.
condition = ["after 10 min of active phone use", "after 10 min without touching phone"]
empty_hand_rating = [3, 0]   # example data: 0 (normal) to 5 (very strange/empty)

results = pd.DataFrame({"condition": condition, "empty-hand rating (0-5)": empty_hand_rating})
display(results)
results.plot.barh(x="condition", y="empty-hand rating (0-5)", legend=False, figsize=(7, 2.5))
plt.xlim(0, 5); plt.tight_layout(); plt.show()
print("A higher rating after active use than after a phone-free period would be consistent with the phone")
print("having been briefly incorporated into your body schema, the way the stick was in the main Lab.")
```
