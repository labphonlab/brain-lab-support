## From glitch to question

The Final Lab has no Brain Mechanism, AI Connection, or Beyond the Lab section, because those are yours to
write. It teaches you to turn a noticed "glitch" into a question, sharpen the question into a testable
design (one thing changed, one thing measured), run the simplest valid version of that design (blind
yourself if you can, counterbalance the order of conditions, repeat each condition several times), and
then analyse and write up what you found -- honestly, including its limits.

## Plan your experiment

**Your question.** Start from a glitch you have actually noticed (see the book's glitch-journal
exercise): something small and repeatable about your own mind that surprised you. Write it as a plain
question with a clear yes/no or more/less shape -- "Does [condition A] produce a different result from
[condition B] on [measurement X]?"

**Your hypothesis.** State, before you collect any data, which condition you expect to score higher (or
lower) on your measurement, and why -- ideally by naming the brain mechanism you think is responsible,
the way the worked example below traces the Lyric Effect to the phonological loop.

**Your method.** Identify exactly one thing to change (your independent variable) and exactly one thing
to measure (your dependent variable) -- everything else stays fixed. Then apply the Final Lab's three
principles: **blind yourself** (have someone else randomise which condition comes when, so your own
expectation cannot leak into your performance or your rating); **counterbalance the order** (alternate
A-B-A-B, or randomise, so practice or fatigue does not masquerade as your effect); and **repeat** (aim
for five to ten trials per condition -- one trial proves nothing). If your design involves another
person as a participant, get their informed consent, tell them about any deception involved and debrief
them immediately afterwards, and let them withdraw at any time.

## Worked example: the Lyric Effect

Before analysing your own data, work through the chapter's own worked example -- **the Lyric Effect** --
step by step, exactly as it appears in the Final Lab text. This shows what a finished analysis looks
like, including its honest limits, before you apply the same steps to your own study.

```python
# The Lyric Effect: reading time in seconds, three trials per condition (from the book's own table).
lyric_effect = pd.DataFrame({
    "Silence":      [38, 35, 36],
    "Instrumental": [39, 37, 38],
    "Lyrics":       [47, 44, 45],
})
display(lyric_effect)

means = lyric_effect.mean()
display(means.rename("mean reading time (s)"))

means.plot.bar(color=["grey", "steelblue", "firebrick"], figsize=(5, 3))
plt.ylabel("mean reading time (s)"); plt.title("The Lyric Effect (worked example)")
plt.show()

lyrics_effect_s = means["Lyrics"] - means["Silence"]
lyrics_effect_pct = 100 * lyrics_effect_s / means["Silence"]
print(f"Lyrics slowed reading by {lyrics_effect_s:.1f} s versus silence "
      f"({lyrics_effect_pct:.0f}% increase), in this illustrative run.")
print(f"Instrumental music vs. silence: {means['Instrumental'] - means['Silence']:.1f} s difference.")
print()
print("As the book itself says: with only three trials per condition and one reader, this is nowhere")
print("near enough data to be confident the gap is real rather than chance variation in reading speed.")
print("Nine trials tell you what pattern to look for, not that you have proven it. Treat every number")
print("above as an illustration of the analysis steps, not as evidence you should cite.")
```

## An AI research assistant

The Final Lab repeatedly recommends using an AI chatbot as a research partner: to check whether your
glitch has already been studied, to suggest a standard experimental paradigm, to generate matched
stimuli, and to help you think through confounds. Below, a small open instruct model plays that role for
a filled-in example question, so you can see both what it is useful for and where you must check its
output yourself.

*This part downloads a small open model or library, so it runs in Colab (it may take a minute the first time).*

```python
!pip install -q transformers accelerate

from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "Qwen/Qwen2.5-0.5B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, torch_dtype="auto")

# Replace this with your own question from my_study above.
your_question = "Does chewing gum while studying improve recall of a word list compared with not chewing gum?"

research_prompt = (
    f"A student wants to test this question with a simple two-condition experiment: {your_question}\n"
    "Suggest: (1) one independent variable and one dependent variable, (2) one likely confound to "
    "control for, and (3) a simple way to counterbalance the order of the two conditions. Be concise."
)
messages = [{"role": "user", "content": research_prompt}]
text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
inputs = tokenizer([text], return_tensors="pt")
out = model.generate(**inputs, max_new_tokens=200, do_sample=False)
suggestion = tokenizer.decode(out[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True)
print(suggestion.strip())
```

Read the model's suggestion critically before using any of it. A small open model like this one can
confidently propose a confound that isn't real, misremember or invent a "standard paradigm" that does
not exist, or suggest a counterbalancing scheme that doesn't actually control for practice effects -- it
has no way to check its own suggestions against the literature, and it cannot judge whether your
specific plan is ethical or feasible. Use it the way the book suggests: as a fast first draft to react to
and correct, exactly as you would want a first-year research assistant's suggestions checked by someone
more experienced -- not as a source of settled fact.

## Replicate across studies

The book's own test of "what makes it science" is replication: share your write-up with others and see
whether they get the same result. Pool everyone's own two-condition experiments here -- different people
can be testing entirely different questions -- using anonymous IDs, and check only the one thing every
study you pool has in common: did the measurement move in the *predicted* direction (as stated in each
person's own hypothesis)?

```python
import io

csv_text = """id,question,condition_a_mean,condition_b_mean,predicted_direction
P01,lyrics vs silence reading time,36.3,45.3,b_higher
P02,caffeine vs no caffeine reaction time,412,388,b_lower
P03,music vs silence memory recall,7.2,6.1,b_lower
P04,warm vs cool light room focus rating,5.8,7.1,b_higher
"""  # example data: replace with your class's own pooled studies
class_df = pd.read_csv(io.StringIO(csv_text))

def matched_direction(row):
    if row["predicted_direction"] == "b_higher":
        return row["condition_b_mean"] > row["condition_a_mean"]
    return row["condition_b_mean"] < row["condition_a_mean"]

class_df["matched_prediction"] = class_df.apply(matched_direction, axis=1)
display(class_df)

replication_rate = class_df["matched_prediction"].mean()
boot = [class_df["matched_prediction"].sample(len(class_df), replace=True).mean() for _ in range(5000)]
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f"{class_df['matched_prediction'].sum()} of {len(class_df)} independent studies moved in their own")
print(f"predicted direction ({replication_rate:.0%}, 95% bootstrap interval {lo:.0%}-{hi:.0%}).")
print("This does not tell you whether any single study is correct -- each used its own design, sample")
print("size, and measurement -- but a group of unrelated glitches that mostly point the predicted way is")
print("a weak, informal echo of what a real replication programme checks formally.")
```
